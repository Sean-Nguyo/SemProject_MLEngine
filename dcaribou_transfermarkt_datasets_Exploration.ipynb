```python
# Google Colab Notebook: Transfermarkt Dataset Analysis
import pandas as pd
url = "https://pub-e682421888d945d684bcae8890b0ec20.r2.dev/data/players.csv.gz"

df = pd.read_csv(url, compression='gzip')

print("First 5 rows of the dataset:")
print(df.head())
```

```python
rows, cols = df.shape
print(f"The dataset contains {rows} rows and {cols} columns.")
```

```python
print("Data types of each column:")
print(df.dtypes)
```

```python
missing_values = df.isnull().sum()
print("Missing values per column:")
print(missing_values)

#total missing values
total_missing = missing_values.sum()
print(f"\nTotal missing values in the dataset: {total_missing}")
if total_missing == 0:
    print("The dataset is complete (no missing values).")
else:
    print("The dataset contains missing values.")
```

```python
first_15 = df.head(15)
last_20 = df.tail(20)

# Merge the two slices vertically
df_sample = pd.concat([first_15, last_20], ignore_index=True)

print("df_sample (first 15 rows + last 20 rows):")
print(df_sample)
```
